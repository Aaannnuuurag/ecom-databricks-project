```python
dbutils.widgets.text("catalog", "anurag_dev")
catalog = dbutils.widgets.get("catalog")

from pyspark.sql.functions import broadcast
from pyspark.sql import functions as F

orders = spark.table(f"{catalog}.silver.orders_enriched")
products = spark.table(f"{catalog}.bronze.products")    # small table → broadcast

# BROADCAST JOIN (Task 4)
result = orders.join(broadcast(products), "product_id")

# REPARTITION vs COALESCE
orders_repartitioned = orders.repartition(8, "state")   # redistribute evenly
orders_coalesced = orders.coalesce(2)                    # reduce without shuffle

print("✅ Optimization operations complete")
```
